```python
from typing import List, Dict, Tuple

def read_processed_dialogues(file_path: str) -> List[Dict[str, str]]:
    """
    Read processed dialogues from a tab-separated text file.
    
    Args:
        file_path (str): Path to the processed dialogues file
        
    Returns:
        List[Dict[str, str]]: List of dictionaries containing dialogue information
            Each dictionary has keys: 'dataset', 'split', 'sequence'
    """
    dialogues = []
    
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            # Split line by tabs and strip whitespace
            parts = [part.strip() for part in line.split('\t')]
            
            if len(parts) == 3:
                dialogue = {
                    'dataset': parts[0],
                    'split': parts[1],
                    'sequence': parts[2]
                }
                dialogues.append(dialogue)
    
    return dialogues

def parse_dialogue_turns(sequence: str) -> List[Tuple[str, str]]:
    """
    Parse a dialogue sequence into separate user and system turns.
    
    Args:
        sequence (str): The dialogue sequence with [USR] and [SYS] tokens
        
    Returns:
        List[Tuple[str, str]]: List of (speaker, utterance) tuples
    """
    turns = []
    # Split by tokens while keeping the tokens
    parts = sequence.replace('[USR]', '\n[USR]').replace('[SYS]', '\n[SYS]').split('\n')
    
    for part in parts:
        part = part.strip()
        if part:
            if part.startswith('[USR]'):
                speaker = 'user'
                utterance = part[5:].strip()  # Remove [USR] token
            elif part.startswith('[SYS]'):
                speaker = 'system'
                utterance = part[5:].strip()  # Remove [SYS] token
            else:
                continue
                
            if utterance:  # Only add if there's actual content
                turns.append((speaker, utterance))
    
    return turns

def get_dialogues_by_dataset(dialogues: List[Dict[str, str]], dataset_name: str) -> List[Dict[str, str]]:
    """
    Filter dialogues by dataset name.
    
    Args:
        dialogues (List[Dict[str, str]]): List of dialogue dictionaries
        dataset_name (str): Name of the dataset to filter by
        
    Returns:
        List[Dict[str, str]]: Filtered list of dialogues from the specified dataset
    """
    return [d for d in dialogues if d['dataset'] == dataset_name]

def get_dialogues_by_split(dialogues: List[Dict[str, str]], split: str) -> List[Dict[str, str]]:
    """
    Filter dialogues by split type (train/dev/test).
    
    Args:
        dialogues (List[Dict[str, str]]): List of dialogue dictionaries
        split (str): Split type to filter by ('train', 'dev', or 'test')
        
    Returns:
        List[Dict[str, str]]: Filtered list of dialogues from the specified split
    """
    return [d for d in dialogues if d['split'] == split]

    
```

```python
dialogues = read_processed_dialogues('processed_dialogues.txt')

# Example: Get all MultiWOZ training dialogues
multiwoz_train = get_dialogues_by_dataset(
    get_dialogues_by_split(dialogues, 'train'), 
    'multiwoz'
)

# Example: Parse turns from first dialogue
if multiwoz_train:
    turns = parse_dialogue_turns(multiwoz_train[0]['sequence'])
    print(turns)
    for speaker, utterance in turns:
        print(f"{speaker}: {utterance}") 
```
